```python
from evaluation import run_evaluation, summarize_results
import pandas as pd
```

```python
df_vanilla = run_evaluation("vanilla", "base")
df_res_1 = run_evaluation("resampling_jump_length_10_jump_n_samples_10", "base")
df_res_2 = run_evaluation("resampling_jump_length_5_jump_n_samples_5", "base")
df_res_3 = run_evaluation("resampling_cfg=0.5_jump_length_10_jump_n_samples_10", "base")
df_res_4 = run_evaluation("resampling_cfg=2_jump_length_10_jump_n_samples_10", "base")
df_res_5 = run_evaluation("resampling_cfg=0.5_jump_length_5_jump_n_samples_5", "base")
df_res_6 = run_evaluation("resampling_cfg=2_jump_length_5_jump_n_samples_5", "base")
```

```python
exp_dfs = [df_vanilla, df_res_1, df_res_2, df_res_3, df_res_4, df_res_5, df_res_6]
exp_names = ["vanilla", "res_10_10", "res_5_5", "res_cfg_0.5", "res_cfg_2", "res_cfg_0.5_5_5", "res_cfg_2_5_5"]

summary_data = summarize_results(exp_dfs, exp_names)
df_all_experiments = summary_data["all_results"]

df_all_experiments.to_csv("all_in_one_results.csv", index=False)

```

```python
metrics = ['PSNR', 'SSIM', 'LPIPS']
overall_df = summary_data['raw_means']['by_image'].groupby('exp')[metrics].mean().T

print("=== OVERALL EXPERIMENT AVERAGES ===")
display(overall_df)
```

```python
print("=== DETAILED BREAKDOWNS ===")
tables = summary_data['tables']

for title, df in tables.items():
    metric = title.split('_')[0]
    is_higher_better = metric in ['PSNR', 'SSIM']
    
    styled = df.style.highlight_max(axis=1, color='lightgreen' if is_higher_better else 'coral') \
                     .highlight_min(axis=1, color='coral' if is_higher_better else 'lightgreen') \
                     .format("{:.4f}")
    
    print(f"Table: {title}")
    display(styled)
    print("-" * 30)
```
