```python
from prepare_images import load_objects_from_json
import matplotlib.pyplot as plt
from PIL import Image
from test_runner import run_vanilla_tests
from masks import create_chessboard_grid_mask, create_col_slit_grid_mask, create_row_slit_grid_mask, create_stochastic_mask
```

```python
images = load_objects_from_json("images_metadata.json","project_images")
masks = []
```

CHESSBOARD

```python
for k in range(1,9):
    mask = create_chessboard_grid_mask(k)
    print(mask.name)
    plt.imshow(mask.image, cmap='gray')
    plt.axis('off') # This removes the axis numbers
    plt.show()

    masks.append(mask)
```

SLIT

```python
for k in range(1,9):
    mask = create_col_slit_grid_mask(k, k)
    print(mask.name)
    plt.imshow(mask.image, cmap='gray')
    plt.axis('off') # This removes the axis numbers
    plt.show()
    masks.append(mask)
```

```python
for k in range(1,8):
    mask = create_row_slit_grid_mask(k, k)
    print(mask.name)
    plt.imshow(mask.image, cmap='gray')
    plt.axis('off') # This removes the axis numbers
    plt.show()
    masks.append(mask)
```

STOCHASTIC

```python
for i in range(0,3):
    mask = create_stochastic_mask(0.5, seed = 40+i)
    mask.name = f"{mask.name}_i={i}"
    print(mask.name)
    plt.imshow(mask.image, cmap='gray')
    plt.axis('off') # This removes the axis numbers
    plt.show()
    masks.append(mask)
```
